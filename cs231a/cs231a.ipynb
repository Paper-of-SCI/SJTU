(Lens Camera)

\begin{equation}

\mathbf{P}'

= 

\begin{bmatrix} x' \\ y' \end{bmatrix}

=

\begin{bmatrix} z'\frac{x}{z} \\ z'\frac{y}{z} \end{bmatrix}

\tag{1}

\end{equation}

### 3D world to digital image coordinates 
(Pinhole Camera)


\begin{equation}

\mathbf{P}' 

= 

\begin{bmatrix} x' \\ y ' \end{bmatrix}

= 

\begin{bmatrix} f\frac{x}{z}+C_x \\ f\frac{y}{z}+C_y   \end{bmatrix}

\tag{2}

\end{equation}


> 从当前往前内容的`Units`均为`Centimeters`。 <br />
>现在要考虑的是对于数字图像平面的离散像素情况，如何去变换单位。接下来将引入`k`和`l`两个参数去解决这个问题。

$ x $ and $ y $ units is $\frac{pixels}{cm}$ (即每厘米有多少像素)

对应$ x $ and $ y $不一定相等， 即一个像素的纵横比不一定是1。如果他俩相等则叫正方形像素。

Digital image coordinates 

$ \mathbf{P} \to \mathbf{P'} $


\begin{equation}

\mathbf{P}' 

= 

\begin{bmatrix} x' \\ y ' \end{bmatrix}

= 

\begin{bmatrix} fk\frac{x}{z}+C_x \\ fl\frac{y}{z}+C_y   \end{bmatrix}

=

\begin{bmatrix} \alpha\frac{x}{z}+C_x \\ \beta\frac{y}{z}+C_y   \end{bmatrix}

\tag{3}

\end{equation}


---
### Better way to represent this projection from $ \mathbf{P}\to \mathbf{P}' $

> 如果一个投影是线性变换，那么它就可以表示为一个矩阵与输入向量（此处输入向量为$\mathbf{P}$）的乘积。对于`Equation 3` 我们知道这个变换是非线性的，因为运算中要对分量`z`做除法。尽管如此，把这种投影写成矩阵与向量的乘积，对后续推导会很有用。<br />
> 乘以矩阵 + 向量加法（加偏置）$\to$ 仿射，可用齐次坐标变成矩阵乘法。
