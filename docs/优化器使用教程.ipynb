```python
import torch
from torch import nn

x = nn.Parameter(torch.tensor([2.0]))

optimizer = torch.optim.SGD([x], lr=0.1)

loss = (x - 5.0) ** 2

loss.backward()

optimizer.step()

print(x)
```

```python
import torch
from torch import nn

x = nn.Parameter(torch.tensor([2.0]))
optimizer = torch.optim.SGD([x], lr=0.1)

for step in range(5):
    optimizer.zero_grad()

    loss = (x - 5.0) ** 2
    loss.backward()
    print(f"本次更新前x: {x.item()}")
    print(f"梯度: {x.grad}")
    optimizer.step()

    print("步数：",step, "更新后x =", x.item(), "loss =", loss.item())
    print("-" * 30)
```
